```python

# ============================================================
# PROJECT: MUSIC DATA ANALYSIS (SPOTIFY DATASET)
# ============================================================

# STEP 1: INSTALL AND IMPORT LIBRARIES

!pip install pandas numpy matplotlib seaborn scikit-learn -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

warnings.filterwarnings('ignore')
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")


# ============================================================
# STEP 2: LOAD DATASET
# ============================================================

# For Google Colab, upload your CSV file first.
from google.colab import files

uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("\nDataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst Five Rows:")
display(df.head())


# ============================================================
# STEP 3: UNDERSTANDING THE DATASET
# ============================================================

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nTotal Genres:", df['track_genre'].nunique())

print("\nTotal Artists:", df['artists'].nunique())


# ============================================================
# STEP 4: DATA CLEANING
# ============================================================

# Remove unnecessary index column
if 'Unnamed: 0' in df.columns:
    df.drop(columns=['Unnamed: 0'], inplace=True)

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Remove rows with missing important values
df.dropna(
    subset=['track_name', 'artists', 'track_genre', 'popularity'],
    inplace=True
)

# Convert duration into minutes
df['duration_min'] = df['duration_ms'] / 60000

print("\nData Cleaning Completed!")
print("Cleaned Dataset Shape:", df.shape)

display(df.head())


# ============================================================
# STEP 5: SONG POPULARITY ANALYSIS
# ============================================================

# Find top 10 popular songs
top_songs = df.sort_values(
    by='popularity',
    ascending=False
).head(10)

print("\nTOP 10 POPULAR SONGS:")
display(top_songs[
    ['track_name', 'artists', 'track_genre', 'popularity']
])

# Average popularity
print("\nAverage Song Popularity:",
      round(df['popularity'].mean(), 2))

# Correlation with popularity
audio_features = [
    'danceability',
    'energy',
    'loudness',
    'speechiness',
    'acousticness',
    'instrumentalness',
    'liveness',
    'valence',
    'tempo',
    'duration_ms'
]

correlation = df[
    audio_features + ['popularity']
].corr()

print("\nCorrelation With Popularity:")
display(
    correlation[['popularity']].sort_values(
        by='popularity',
        ascending=False
    )
)


# ============================================================
# STEP 6: POPULARITY DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    df['popularity'],
    bins=30,
    kde=True
)

plt.title("Distribution of Song Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Songs")
plt.show()


# ============================================================
# STEP 7: CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Between Music Features")
plt.show()


# ============================================================
# STEP 8: DANCEABILITY VS POPULARITY
# ============================================================

plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x='danceability',
    y='popularity',
    alpha=0.4
)

plt.title("Danceability vs Song Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity")
plt.show()


# ============================================================
# STEP 9: GENRE ANALYSIS
# ============================================================

# Average popularity of each genre
genre_popularity = df.groupby(
    'track_genre'
)['popularity'].mean().sort_values(ascending=False)

print("\nTOP 10 GENRES BY POPULARITY:")
display(genre_popularity.head(10))

# Visualize top 10 genres
plt.figure(figsize=(12, 6))

genre_popularity.head(10).plot(
    kind='bar',
    color='green'
)

plt.title("Top 10 Music Genres by Average Popularity")
plt.xlabel("Music Genre")
plt.ylabel("Average Popularity")
plt.xticks(rotation=45)
plt.show()


# ============================================================
# STEP 10: COMPARE AUDIO FEATURES ACROSS GENRES
# ============================================================

genre_features = df.groupby('track_genre')[
    ['danceability', 'energy', 'valence', 'acousticness']
].mean()

# Select 10 genres with highest average popularity
selected_genres = genre_popularity.head(10).index

selected_features = genre_features.loc[selected_genres]

selected_features.plot(
    kind='bar',
    figsize=(14, 6)
)

plt.title("Audio Feature Comparison Across Genres")
plt.xlabel("Music Genre")
plt.ylabel("Average Feature Value")
plt.xticks(rotation=45)
plt.legend(title="Audio Features")
plt.show()


# ============================================================
# STEP 11: TRENDING SONG ANALYSIS
# ============================================================

# Define popularity categories
df['popularity_category'] = pd.cut(
    df['popularity'],
    bins=[-1, 30, 70, 100],
    labels=['Low', 'Medium', 'High']
)

print("\nPopularity Category Counts:")
print(df['popularity_category'].value_counts())

# Compare audio features
popularity_analysis = df.groupby(
    'popularity_category',
    observed=False
)[[
    'danceability',
    'energy',
    'valence',
    'acousticness',
    'tempo'
]].mean()

print("\nPopularity Category Analysis:")
display(popularity_analysis)

# Visualize
popularity_analysis.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title("Audio Features Across Popularity Categories")
plt.xlabel("Popularity Category")
plt.ylabel("Average Feature Value")
plt.xticks(rotation=0)
plt.show()


# ============================================================
# STEP 12: EXPLICIT VS NON-EXPLICIT SONG ANALYSIS
# ============================================================

explicit_analysis = df.groupby('explicit')[
    ['popularity', 'danceability', 'energy', 'valence']
].mean()

print("\nExplicit vs Non-Explicit Song Analysis:")
display(explicit_analysis)

explicit_analysis.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title("Explicit vs Non-Explicit Songs")
plt.xlabel("Explicit (False = No, True = Yes)")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.show()


# ============================================================
# STEP 13: K-MEANS CLUSTERING
# ============================================================

cluster_features = [
    'danceability',
    'energy',
    'loudness',
    'speechiness',
    'acousticness',
    'instrumentalness',
    'liveness',
    'valence',
    'tempo'
]

# Prepare clustering data
X = df[cluster_features].copy()

X = X.replace([np.inf, -np.inf], np.nan)
X = X.fillna(X.median())

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Select a sample for efficient model fitting
sample_size = min(20000, len(X_scaled))

rng = np.random.RandomState(42)

sample_indices = rng.choice(
    len(X_scaled),
    size=sample_size,
    replace=False
)

X_sample = X_scaled[sample_indices]

# Apply K-Means with 4 clusters
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

kmeans.fit(X_sample)

# Predict cluster for all songs
df['cluster'] = kmeans.predict(X_scaled)

print("\nK-Means Clustering Completed!")

print("\nSongs in Each Cluster:")
print(df['cluster'].value_counts().sort_index())


# ============================================================
# STEP 14: CLUSTER VISUALIZATION
# ============================================================

plot_df = df.sample(
    min(5000, len(df)),
    random_state=42
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=plot_df,
    x='danceability',
    y='energy',
    hue='cluster',
    palette='viridis',
    alpha=0.6
)

plt.title("Spotify Songs Grouped into Clusters")
plt.xlabel("Danceability")
plt.ylabel("Energy")
plt.show()


# ============================================================
# STEP 15: CLUSTER ANALYSIS
# ============================================================

cluster_analysis = df.groupby('cluster')[
    cluster_features + ['popularity']
].mean()

print("\nCLUSTER CHARACTERISTICS:")
display(cluster_analysis)

# Top 5 popular songs from each cluster
for cluster_num in sorted(df['cluster'].unique()):

    print(f"\nTOP SONGS IN CLUSTER {cluster_num}:")

    display(
        df[df['cluster'] == cluster_num]
        .sort_values(
            by='popularity',
            ascending=False
        )[
            ['track_name', 'artists', 'popularity']
        ].head(5)
    )


# ============================================================
# STEP 16: SILHOUETTE SCORE
# ============================================================

score = silhouette_score(
    X_sample,
    kmeans.labels_
)

print("\nSilhouette Score:", round(score, 4))


# ============================================================
# STEP 17: FINAL OBSERVATIONS
# ============================================================

print("\n========== PROJECT SUMMARY ==========")

print("Total Songs Analyzed:", len(df))

print("Total Genres:", df['track_genre'].nunique())

print("Total Artists:", df['artists'].nunique())

print(
    "Average Popularity:",
    round(df['popularity'].mean(), 2)
)

print(
    "Most Popular Genre:",
    genre_popularity.index[0]
)

print(
    "Highest Popularity Score:",
    df['popularity'].max()
)

print(
    "Number of Clusters:",
    df['cluster'].nunique()
)

print("\nProject Analysis Completed Successfully!")


# ============================================================
# STEP 18: SAVE FINAL RESULTS
# ============================================================

# Save analyzed dataset
df.to_csv(
    "spotify_analysis_results.csv",
    index=False
)

# Save genre analysis
genre_popularity.to_csv(
    "genre_popularity.csv"
)

# Save cluster analysis
cluster_analysis.to_csv(
    "cluster_analysis.csv"
)

print("\nAll Result Files Saved Successfully!")


# ============================================================
# STEP 19: DOWNLOAD RESULTS (GOOGLE COLAB)
# ============================================================

files.download("spotify_analysis_results.csv")
files.download("genre_popularity.csv")
files.download("cluster_analysis.csv")

print("\nALL TASKS COMPLETED SUCCESSFULLY!")
```
